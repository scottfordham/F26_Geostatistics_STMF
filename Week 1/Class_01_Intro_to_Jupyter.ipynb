# Before we Begin

## Python, Jupyter, and the kernel

Before writing code, it helps to separate three ideas that are easy to mix up:

### Python
**Python** is the programming language. A Python interpreter is the program that reads and executes Python code.

### Jupyter Notebook
**Jupyter Notebook** is the interactive interface we use to combine code, text, equations, figures, and results.

### The kernel
The **kernel** is the running computational process behind the notebook.

When you run a Python code cell, Jupyter sends that code to the Python kernel. The kernel executes it, remembers variables you have created, and sends the result back to the notebook.

A useful mental model is:

> **Notebook = the document and interface**  
> **Kernel = the live Python session doing the computation**

This distinction matters because the *same notebook file* can sometimes be connected to different Python installations or environments.

In Summary:

| Term | What it is |
|---|---|
| **Notebook** | The document itself. Saved on disk as a `.ipynb` file (which is really just a text file in JSON format). |
| **Kernel** | The Python process running behind the notebook. It holds all your variables in memory and executes the code you send it. |


````python
# Ask the current kernel which Python it is using

import sys
import platform

print("Python version:")
print(sys.version)

print("\nPython location:") # the Python program file that this notebook’s kernel is using
print(sys.executable)

print("\nOperating system:")
print(platform.platform())
````

### Why kernel state matters

The kernel remembers variables only while that computational session is running.

For example, if you run:

```python
x = 10
```

the kernel remembers `x`.

If you **restart the kernel**, that memory is cleared. The notebook file still contains the code, but Python no longer remembers the variables until you run the relevant cells again.

This explains a very common beginner experience:

> “But the code is right there in the notebook, why does Python say the variable does not exist?”

Usually, the cell that created the variable has not been run in the current kernel session.

### A reproducibility habit

A notebook should ideally work when run from top to bottom in a fresh kernel.

During the semester, a good check is:

**Restart Kernel → Run All Cells**

If the notebook still works, your analysis is much more likely to be reproducible.

For this course, **Python is already available on the laboratory machines**, so you do not need to install anything to participate in class.

Outside the lab, however, there are several ways to use Python. There is no single required setup for your personal computer.

## Option 1 — Conda: recommended for most students

For most students who want Python on their own computer, **Conda is the easiest local option I recommend**.

Conda helps manage both:

A local environment might look conceptually like this:


The official Conda documentation recommends a minimal installer such as **Miniconda** or **Miniforge** for getting started.

- [Conda documentation; Getting Started](https://docs.conda.io/projects/conda/en/latest/user-guide/getting-started.html)

Email me if you need installation help or you could come to office hours.


## Option 2 — Google Colab: no local installation

**Google Colab** runs Jupyter-style notebooks in a web browser.

This is useful when:

- you are away from the laboratory,
- you are using a computer where you cannot install software,
- you want to experiment quickly,
- or you want to share a notebook easily.

Google describes Colab as a hosted Jupyter Notebook service that requires no local setup.

- [Open Google Colab](https://colab.research.google.com/)
- [Google Colab FAQ](https://research.google.com/colaboratory/faq.html)

### One important difference

With a local installation, the Python kernel runs **on your computer**.

With standard Google Colab, the Python runtime/kernel generally runs **on Google's computing infrastructure**.

That means a notebook can look nearly identical while the computation is happening somewhere completely different.


## Option 3 — pyenv: useful, but more advanced

Another popular tool is **pyenv**.

pyenv is particularly useful when you want fine control over **which version of Python** is installed and active on your machine.

For example, a developer might need one project to use one Python version and another project to use a different version.

- [pyenv project and installation instructions](https://github.com/pyenv/pyenv)

For this class, you **do not need pyenv**.

It is a powerful tool, but it generally requires more comfort with the command line and system configuration. If your goal is simply to install Python, Jupyter, and the scientific Python packages used in this course, **Conda is usually the easier starting point**.

# What is Jupyter Notebook?

A **Jupyter Notebook** is an interactive document that can contain:

- formatted text,
- equations,
- Python code,
- tables,
- plots,
- analysis results.

This makes notebooks especially useful for **reproducible data analysis**: the explanation, code, figures, and interpretation can all live in the same document.

The notebook interface communicates with the **Python kernel** introduced above.

A notebook is made of **cells**. The two cell types we will use most often are:

### Markdown cells
Used for headings, explanations, equations, instructions, and interpretation.

### Code cells
Used to execute Python code.

Three ways to run the selected cell:

| Shortcut | What it does |
|---|---|
| `Shift` + `Enter` | Run the cell and move to the next one. *(Your default.)* |
| `Ctrl` + `Enter` | Run the cell and stay on it. *(Good for re-running while tweaking.)* |
| `Alt` + `Enter` | Run the cell and insert a new empty cell below. |

On a Mac, `Ctrl` is `Control` and `Alt` is `Option`.



````python
print("Welcome to Data Analysis and Geostatistics!")
````

### Execution order matters

When a code cell runs, Jupyter stores the result in the current Python session.

For example, run the next two cells **in order**.


````python
course_name = "Data Analysis and Geostatistics"
print(course_name)
````

````python
print("This course is:", course_name)
````

If you restart the notebook kernel, stored variables disappear until the cells that define them are run again.

A good habit is to periodically use:

**Kernel → Restart Kernel and Run All Cells**

This checks whether your notebook runs correctly from beginning to end.
